# Display Data

## Load display package

```python
# Import packages for display to streamlit charts
import numpy as np
import streamlit as st
import plotly.express as px
```

## Visualisations

```python
# Display pie chart of gender distribution
category_counts = police_df['gender'].value_counts().reset_index()
fig = px.pie(category_counts, names='gender', values='count', title='Gender Distribution')
fig.show()
```

```python
# Print list of unique age range values in police_df
print(police_df['age_range'].unique())
```

```python
# Create a bar chart of age distribution and order by provided age order
age_order = ('Under 10', '10-17', '18-24', '25-34', 'over 34', 'Not Recorded')
age_counts = police_df['age_range'].value_counts().reindex(age_order).reset_index()
fig = px.bar(age_counts, x='age_range', y='count', title='Age Distribution', labels={'index': 'Age Group', 'count': 'Number of Incidents'})
fig.show()
```

```python
# Use previous bar chart and add colour to differentiate gender, keeping the previous ordering of age ranges
age_gender_counts = police_df.groupby(['age_range', 'gender']).size().reindex(age_order, level=0).reset_index(name='count')
fig = px.bar(age_gender_counts, x='age_range', y='count', color='gender', title='Age Distribution by Gender', labels={'age_range': 'Age Group', 'count': 'Number of Incidents'})
fig.show()
```
